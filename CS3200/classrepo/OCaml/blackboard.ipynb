#### Structural recursion vs. generative (non-structural) recursion.

Structural recursion uses a subset of the original data. Generative recursion uses data that was generated based on the original data.
	
Generatively recursive functions may diverge on some inputs, but not structurally recursive functions (because a subset is always smaller than the original and the size guaranteed to reach 0.)

```ocaml
let rec gcd a b = (* greatest common divisor *)
        if b = 0 then a else gcd b (a mod b);;
```

```ocaml
gcd 24 60;;
gcd 6 4;;
```

### PA2: Natural numbers and binary-encoded numbers 

```ocaml
type pos =
  | H
  | O of pos
  | I of pos

(** The type 'pos' has three constructors, 'H', 'O', and 'I'. They can
   be understood as follows:

   * 'H' stands for 1.
   * 'O n', the 'O' constructor applied to a pos n, stands for 2 * n.
   * 'I n', the 'I' constructor applied to a pos n, stands for 2 * n + 1.

   This interpretation of pos is codified by the function 'int_of_pos': *)
```

```ocaml
let one   = H (* 001 *)
let two   = O H  (* 010 *)
let three = I H  (* 011 *)
let four  = O two (* 100 = O O H *)
let five  = I two (* 101 = I O H *)
let six   = O (I H) (* 110 = O I H *)
```

### Codewars: Find the smallest power higher than a given a value

https://www.codewars.com/kata/56ba65c6a15703ac7e002075

```ocaml
let rec power x = function
    | 0 -> 1
    | 1 -> x
    | n -> 
      let y = power x (n / 2) in
        y * y * (if n mod 2 = 0 then 1 else x)
```

```ocaml
power 2 10;;
power 1 10;;
```

```ocaml
let rec find_next (val_: int) (pow: int) (curr: int) : int = if power curr pow > val_ then curr else find_next val_ pow (curr+1)

let find_next_power (val_: int) (pow: int): int = 
    find_next val_ pow 1
```

```ocaml
find_next 10 2 1;;
find_next 20 2 1;;
find_next 100 2 1;;
```

```ocaml
find_next_power 10 2;;
find_next_power 10 3;;
```

### Binary Search Tree

```ocaml
type 'a tree = Node of 'a * 'a tree * 'a tree | Leaf

(** [mem x t] is [true] iff [x] is a member of [t]. *)
let rec mem x = function
  | Leaf -> false
  | Node (y, l, r) ->
    if x < y then mem x l
    else if x > y then mem x r
    else true

(** [insert x t] is [t] . *)
let rec insert x = function
  | Leaf -> Node (x, Leaf, Leaf)
  | Node (y, l, r) as t ->
    if x < y then Node (y, insert x l, r)
    else if x > y then Node (y, l, insert x r)
    else t
```

```ocaml
let bst1 = insert 1 Leaf;;
let bst2 = insert 2 bst1;;
let bst3 = insert 3 bst2;;
let bst4 = insert 4 bst3;;

(* let bst5 =  insert 3 (insert 1 (insert 2 (insert 4 Leaf)));; *)
```

### Red black Tree

```ocaml
type color = Red | Black
type 'a rbtree = Leaf | Node of color * 'a * 'a rbtree * 'a rbtree
```

```ocaml
let rec mem x = function
  | Leaf -> false
  | Node (_, y, l, r) ->
    if x < y then mem x l
    else if x > y then mem x r
    else true
```

```ocaml
let balance = function
  | Black, z, Node (Red, y, Node (Red, x, a, b), c), d
  | Black, z, Node (Red, x, a, Node (Red, y, b, c)), d
  | Black, x, a, Node (Red, z, Node (Red, y, b, c), d)
  | Black, x, a, Node (Red, y, b, Node (Red, z, c, d)) ->
    Node (Red, y, Node (Black, x, a, b), Node (Black, z, c, d))
  | a, b, c, d -> Node (a, b, c, d)
```

```ocaml
let insert x s =
  let rec ins = function
    | Leaf -> Node (Red, x, Leaf, Leaf)
    | Node (color, y, a, b) as s ->
      if x < y then balance (color, y, ins a, b)
      else if x > y then balance (color, y, a, ins b)
      else s
  in
  match ins s with
  | Node (_, y, a, b) -> Node (Black, y, a, b)
  | Leaf -> (* guaranteed to be nonempty *)
    failwith "RBT insert failed with ins returning leaf"
```

```ocaml
let rbt1 = insert 1 Leaf;;
let rbt2 = insert 2 rbt1;;
let rbt3 = insert 3 rbt2;;
let rbt4 = insert 4 rbt3;;
let rbt5 = insert 5 rbt4;;
let rbt6 = insert 6 rbt5;;
let rbt7 = insert 7 rbt6;;

(* let rbt5 =  insert 3 (insert 1 (insert 2 (insert 4 Leaf)));; *)

(* let rbt5 =  insert 3 (insert 1 (insert 2 (insert 4 Leaf)));; *)
```

### Lazy evaluation

Lazy expressions

The expression lazy expr returns a value v of type Lazy.t that encapsulates the computation of expr. The argument expr is not evaluated at this point in the program. Instead, its evaluation will be performed the first time the function Lazy.force is applied to the value v, returning the actual value of expr. Subsequent applications of Lazy.force to v do not evaluate expr again. Applications of Lazy.force may be implicit through pattern matching (see 8.3).

https://v2.ocaml.org/releases/4.14/htmlman/expr.html

```ocaml
module type Lazy2 = sig
    type 'a t = 'a lazy_t
    val force : 'a t -> 'a
  end
```

```ocaml
module Lazy2 : Lazy2 = struct
  type 'a t = 'a lazy_t
  let force = Lazy.force
end
```

```ocaml
let x : int Lazy.t = lazy 1;;
let y : int Lazy.t = lazy 2;;
let z : int Lazy.t = lazy 3;;

Lazy.force x;;
Lazy.force x;;
Lazy.force y;;
```

```ocaml
open Lazy
```

```ocaml
let u: unit Lazy.t = lazy (print_endline "hello lazy world");;
Lazy.force u;;
Lazy.force u;;
Lazy.force u;;
```

### Module Result

* https://v2.ocaml.org/api/Result.html
* Definition of `let*` and `let+` (Result.bind and Result.map) https://github.com/OU-CS3560/pa4-parser-drchangliu/blob/main/lib/util.ml

```ocaml
open Result

let parse_int (s: string): (int, string) Result.t =
  try
    Ok (int_of_string s)
  with
    | _ -> Error "Not a valid integer"

```

```ocaml
let () =
  let parsed_int = parse_int "123" in
  match parsed_int with
    | Ok i -> print_int i; print_newline ();
    | Error s -> print_endline s;
  flush_all();;
```

```ocaml
let () =
  let parsed_int = parse_int "abc" in
  match parsed_int with
    | Ok i -> print_int i; print_newline ();
    | Error s -> print_endline s;
  flush_all();;
```

```ocaml
open Result

let parse_int_list (strs: string list) : (int, string) Result.t list =
  List.map parse_int strs

let print_int_res (res: (int, string) Result.t) : unit =
  match res with
    | Ok i -> print_int i; print_newline ();
    | Error s -> print_endline s;
  flush_all();;

let x =
  let parsed_ints = parse_int_list ["123"; "456"; "789"] in
  List.map print_int_res parsed_ints 
```

```ocaml
(* Define a custom error type *)
type custom_error = 
  | NotFound
  | InvalidInput of string

(* Define a function that may fail *)
let divide x y = 
  if y = 0 then Error (InvalidInput "Division by zero")
  else Ok (x / y)

(* Pattern match on the result *)
let handle_result res =
  match res with
  | Ok result -> Printf.printf "Result: %d\n" result
  | Error error ->
      (match error with
       | NotFound -> Printf.printf "Error: Not found\n"
       | InvalidInput msg -> Printf.printf "Error: Invalid input - %s\n" msg)
```

```ocaml
(* Example usage *)
let () =
  handle_result (divide 10 2);
  handle_result (divide 10 0);

flush_all();;
```

```ocaml
let result1 = divide 6 3 in
handle_result result1;

flush_all();;
```

```ocaml
let result2 = divide 10 0 in
  handle_result result2;

flush_all();;
```
