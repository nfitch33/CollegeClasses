## An auxilary function `tsize` that takes an extra parameter `acc` to remember the accumulative results so far

```ocaml
let rec tsize x acc = 
    match x with
    | [] -> acc + 0
    | h :: t ->  tsize t (acc+1)
```

```ocaml
tsize lst 0
```

```ocaml
let size x = tsize x 0
```

```ocaml
size lst
```

## The tail-recursive version of `size` and `--`

```ocaml
(* The final tail-recursive solution *)

let size x = 
  let rec tsize x acc = 
    match x with
    | [] -> acc
    | h :: t -> tsize t (acc+1)
  in
      tsize x 0
```

```ocaml
size lst
```

```ocaml
let rec f i j acc = 
    if i > j then 
        acc 
    else 
        f (i + 1)  j (List.append acc [i])

(* This is too slow because adding an element at the end of the list is linear to the length of the list. *)

let rec ( -- ) i j = f i j []
```

```ocaml
let c = 1 -- 1000000   
```

```ocaml
let rec f i j acc = 
    if i > j then 
        acc 
    else 
        f (i)  (j-1) (j :: acc)
        
```

```ocaml
let ( -- ) i j = 
    f i j []
```

```ocaml
(* The final efficient tail-recursive solution. *)

let ( -- ) i j = 
  let rec f i j acc = 
    if i > j then 
        acc 
    else 
        f (i)  (j-1) (j :: acc)
in
    f i j []
```

```ocaml
size lst
```

```ocaml
let c = 1 -- 1000000  
```

```ocaml
(*
 Kata: Back and forth then reverse (list efficiency and tail recursion): 
    https://www.codewars.com/kata/60cc93db4ab0ae0026761232
 *)
 
 (* Version 4 by Bagnalla *)
 
open Printf

let show_list n = 
  List.iter (printf "%d : ") n;
  printf "\n"

let show n = 
  printf "%d \n" n


let rec take n l = if n <= 0 then [] else
  match l with
  | [] -> []
  | x :: xs -> x :: take (n - 1) xs
  
let rec drop n l = if n <= 0 then l else
  match l with
  | [] -> []
  | _ :: xs -> drop (n - 1) xs

let zip l r =
  let rec go l r b acc = (* b is the flag to decide which side goes first *)
    match l, r with
    | [], _ -> r @ acc
    | _, [] -> l @ acc
    | x :: xs, y :: ys -> if b then go xs ys (not b) (x :: y :: acc) else go xs ys (not b) (y :: x :: acc) (* This is tail recursion! *)
  in List.rev @@ go l r false []

let arrange (s : 'a list) : 'a list =
  let l = take ((List.length s + 1) / 2) s in
  show_list l;
  let r = List.rev @@ drop ((List.length s + 1) / 2) s in
  show_list r;
  zip l r
;;

```
