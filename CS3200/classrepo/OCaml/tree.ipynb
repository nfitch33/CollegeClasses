```ocaml
let rec size = function
  | Leaf -> 0
  | Node (_, left, right) -> 1 + size left + size right
```

```ocaml
size t3
```

```ocaml
let rec sum = function
  | Leaf -> 0
  | Node (root, left, right) -> root + sum left + sum right
```

```ocaml
let sum t = 
    let rec aux acc = function
        | Leaf -> acc
        | Node (root, left, right) -> aux (root + acc) left + aux (root + acc) right
    in
    aux 0 t
```

```ocaml
sum t7;;
sum t3;;
```

#### Tree2List and List2Tree

```ocaml
let rec preorder = function
  | Leaf -> []
  | Node (value, left, right) ->   [value] @ preorder left @ preorder right
```

```ocaml
let rec list2tree = function
  | [] -> Leaf
  | x :: xs -> Node (x, list2tree (List.filter (fun y -> y < x) xs), list2tree (List.filter (fun y -> y > x) xs))
```

### Binary Search Tree  

```ocaml
type 'a tree = Node of 'a * 'a tree * 'a tree | Leaf

(** [mem x t] is [true] iff [x] is a member of [t]. *)
let rec mem x = function
  | Leaf -> false
  | Node (y, l, r) ->
    if x < y then mem x l
    else if x > y then mem x r
    else true

(** [insert x t] is [t] . *)
let rec insert x = function
  | Leaf -> Node (x, Leaf, Leaf)
  | Node (y, l, r) as t ->
    if x < y then Node (y, insert x l, r)
    else if x > y then Node (y, l, insert x r)
    else t
```

```ocaml
let bst1 = insert 1 Leaf;;
let bst2 = insert 2 bst1;;
let bst3 = insert 3 bst2;;
let bst4 = insert 4 bst3;;
let bst5 = insert 5 bst4;;
let bst6 = insert 6 bst5;;
let bst7 = insert 7 bst6;;
let bst8 = insert 8 bst7;;
let bst9 = insert 9 bst8;;
let bst10 = insert 10 bst9;;
```

### Red Black Tree, a special type of self-balanced binary search tree

* Every node is either red or black.
* The root node is always black.

* *Local Invariant*: There are no two adjacent red nodes along any path.

* *Global Invariant*: Every path from the root to a leaf has the same number of black nodes. This number is called the black height (BH) of the tree.

These properties ensure that the height of a red-black tree is always O(log n),

The color of an empty tree is black.

```ocaml
type color = Red | Black
type 'a rbtree = Leaf | Node of color * 'a * 'a rbtree * 'a rbtree
```

```ocaml
let rec mem x = function
  | Leaf -> false
  | Node (_, y, l, r) ->
    if x < y then mem x l
    else if x > y then mem x r
    else true
```

```ocaml
let balance = function
  | Black, z, Node (Red, y, Node (Red, x, a, b), c), d
  | Black, z, Node (Red, x, a, Node (Red, y, b, c)), d
  | Black, x, a, Node (Red, z, Node (Red, y, b, c), d)
  | Black, x, a, Node (Red, y, b, Node (Red, z, c, d)) ->
    Node (Red, y, Node (Black, x, a, b), Node (Black, z, c, d))
  | a, b, c, d -> Node (a, b, c, d)
```

```ocaml
let insert x s =
  let rec ins = function
    | Leaf -> Node (Red, x, Leaf, Leaf)
    | Node (color, y, a, b) as s ->
      if x < y then balance (color, y, ins a, b)
      else if x > y then balance (color, y, a, ins b)
      else s
  in
  match ins s with
  | Node (_, y, a, b) -> Node (Black, y, a, b)
  | Leaf -> (* guaranteed to be nonempty *)
    failwith "RBT insert failed with ins returning leaf" (* failwith is a library function. https://ocaml.org/docs/error-handling *)
```
