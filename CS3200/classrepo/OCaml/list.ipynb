### Generate a list

```ocaml
let rec generate_a_list_of_ones (n: int) =
  if  n = 0 then []
  else
    1 :: generate_a_list_of_ones (n-1) 
```

```ocaml
let rec generate_a_list_of_ones n =
  match n with
  | 0 -> []
  | 1 -> [1]
  | x -> 1 :: generate_a_list_of_ones (x-1) 
```

```ocaml
generate_a_list_of_ones 10
```

```ocaml
let rec generate_a_list n f =
  if  n = 0 then []
  else
    f n :: generate_a_list (n-1) f 
```

```ocaml
generate_a_list 10 (fun x -> x)
```

The tail recursive version with an auxiliary function and an accumulator.

```ocaml
let rec generate_a_list_of_ones_aux n acc =
  if  n = 0 then acc
  else
    generate_a_list_of_ones_aux (n-1) (1:: acc) 

let generate_a_list_of_ones_T n = generate_a_list_of_ones_aux n []    
```

```ocaml
generate_a_list_of_ones_T 100000000
```

###  Compute the length of a list

```ocaml
let lst = generate_a_list_of_ones 10
```

```ocaml
let rec length (list): int =
  match list with
  | [] -> 0
  | [_] -> 1
  | _ :: list' -> 1 + (length list')
```

```ocaml
length [2]
```

```ocaml
length(2 :: [])
```

```ocaml
length (2 :: [ 3])
```

```ocaml
let x = 2 :: 3 :: 4 :: [5]
```

```ocaml
let y = [1; 2 ; 3; 4; 5; 6]
```

```ocaml
length  x
```

```ocaml
length y
```

```ocaml
(* Only two cases in `match` are necessary for this recursive function. *)

let rec length (list): int =
  match list with
  | [] -> 0
  | _ :: list' -> 1 + (length list')
```

###  Compute the sum of a list

```ocaml
let rec sum (list): int =
  match list with
  | [] -> 0
  | h :: list' -> h + (sum list')
```

```ocaml
let x = [1; 2; 3]
```

```ocaml
sum x
```

```ocaml
sum([1; 2 ;3])
```

```ocaml
let rec sum_acc list sum_so_far: int =
  match list with
  | [] -> sum_so_far
  | h :: list' -> sum_acc list' (h + sum_so_far)

let sumT list = sum_acc list 0
```

```ocaml
let s = sum([])
let s = sum([1; 2])
let s = sum(2::[3])
let s = sum([1; 2 ;3])
let s = sumT([])
let s = sumT([1; 2])
let s = sumT(2::[3])
let s = sumT([1; 2 ;3])
```

### Duplicate every element in the list

```ocaml
let rec double (list) =
  match list with
  | [] -> []
  | x :: list' ->  x :: x :: double list'

 
```

```ocaml
let rec double_element (list) =
  match list with
  | [] -> []
  | x :: list' ->  x* 2 :: double_element list' 
```

```ocaml
double([1; 2; 3])
```

```ocaml
double_element([1; 2; 3])
```

```ocaml
open Float

let rec powersOfTwo n = 
  if n = 0 then [1]
  else
     (powersOfTwo (n-1))  @ [to_int(2.** (of_int n)) ]
```

```ocaml
let rec powersOfTwo_aux acc n = 
    match n with
    | 0 -> [acc]
    | n -> acc :: powersOfTwo_aux (acc * 2) (n - 1) 
let powersOfTwo n = powersOfTwo_aux 1 n
```

```ocaml
let rec fac n = 
  if n <= 1 then 1
  else n * fac (n-1)
```

```ocaml
let fact : int -> int = fun num ->
     let rec factTR : int -> int -> int = fun n acc ->
          print_endline( (string_of_int n) ^ " " ^ (string_of_int acc));
          if n = 1
          then acc
          else factTR (n-1) (acc * n)
     in
          factTR num 1
(* acc is accumulated factorial value. *)
```

```ocaml
fact(5)
```

Note that in this case all five factTR return the same value 120.

```ocaml
let factorial : big_int -> big_int = fun num ->
     let rec factTR : big_int -> big_int -> big_int = fun n acc ->
          if gt_big_int n (big_int_of_int 0)
          then factTR 
            (sub_big_int n (big_int_of_int 1)) 
            (mult_big_int acc n)
          else acc
     in
          factTR num (big_int_of_int 1)
```

```ocaml
let factorial : int -> big_int = fun num ->
  let rec factTR : int -> big_int -> big_int = fun n acc ->
       if n > 0
       then factTR (n - 1) (mult_big_int acc (big_int_of_int n))
       else acc
  in
       factTR num (big_int_of_int 1)
```
