# 2.3.2 structural equality and physical equality 

* Helpful link: https://stackoverflow.com/questions/13590307/whats-the-difference-between-equal-and-identical-in-ocaml

* Caution: the semantics of equality operators depend on whether the variable is mutable.

```ocaml
42 = 42
```

```ocaml
42 == 42
```

```ocaml
let x = 42 in x == 42
```

```ocaml
let x = 42 in x = 42
```

```ocaml
let x = 42 in
let y = 42 in
x = y
```

```ocaml
let x = 42 in
let y = 42 in
x == y
```

```ocaml
let x = [1; 2] in
let y = [1; 2] in 
x = y
```

```ocaml
let x = [1; 2] in
let y = [1; 2] in 
x == y
```

```ocaml
let x = [1; 2] in
x = [1; 2]
```

```ocaml
let x = [1; 2] in
x == [1; 2]
```
