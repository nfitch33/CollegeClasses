### Mutability and the reference type
https://cs3110.github.io/textbook/chapters/mut/refs.html

Mutability helps with efficiency; immutability helps with reasoning and avoiding human errors.

Aliasing: two refs point to the same memory location.

Ref creation: `ref e`

Ref assignment: `e1 := e2`

Dereference: `!e`

The semicolon operator `;` is used to sequence effects, such as mutating refs.

Evil Secret: `Obj.magic` gets a memory address of a ref.

OCaml uses `weak type variables` to stand for unknown but not polymorphic types. These variables always start with _weak. Although the value stored in a ref cell is permitted to change, the type of that value is not.

```ocaml
let x = ref 0;;
```

```ocaml
!x;;
```

```ocaml
x := 1;;
```

```ocaml
!x;;
```

```ocaml
let x = ref 41;;
let y = ref 42;;
let z = x;;
x := 43;;
let w = !y + !z;;
```

```ocaml
!x;;
!y;;
!z;;
w;;
```

### A counter - The simplest stateful function

```ocaml
let counter = ref 0

let next_val =
  fun () ->
    counter := !counter + 1;
    !counter
```

```ocaml
!counter;;
```

```ocaml
next_val()
```

```ocaml
next_val()
```

```ocaml
next_val()
```

```ocaml
!counter;;
```

### Structural equality versus physical equality

```ocaml
let r1 = ref 42
let r2 = ref 42
```

```ocaml
r1 == r1;;
r1 = r1
```

```ocaml
r1 == r2;;
r1 = r2
```

```ocaml
!r1 = !r2 ;;
!r1 == !r2
```

### Pointers

```ocaml
type 'a pointer = 'a ref option
let null : 'a pointer = None

let malloc (x : 'a) : 'a pointer = Some (ref x)

let p = malloc 42

exception Segfault

let deref (ptr : 'a pointer) : 'a =
  match ptr with None -> raise Segfault | Some r -> !r
```
