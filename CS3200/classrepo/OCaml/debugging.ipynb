### CodeWars: finding multiples: 
https://www.codewars.com/kata/58ca658cc0d6401f2700045f/

```ocaml
open Printf

let rec find_multiples_r (integer: int) (current_integer: int) (limit: int): int list =
    printf "-- %d %d %d\n" integer current_integer limit;
    if limit < current_integer then []
    else current_integer :: (find_multiples_r integer (current_integer+integer) limit)

let find_multiples (integer: int) (limit: int): int list =
    printf "%d %d\n" integer limit;
    find_multiples_r integer integer limit
```

```ocaml
(find_multiples 5 25)
```

```ocaml
find_multiples 2 6 
```

### Abbreviate a Two Word Name
https://www.codewars.com/kata/57eadb7ecd143f4c9c0000a3

```ocaml
open Printf

let show_list n = 
  List.iter (printf "%s : ") n;
  printf "\n";
  n

let show n = 
  printf "%s \n" n;
  n


let abbrev_name name =
  String.split_on_char ' ' name
    |> show_list
    |> List.map (fun s -> String.sub s 0 1)
    |> show_list
    |> String.concat "."
    |> show
    |> String.uppercase_ascii
    |> show
```

```ocaml
abbrev_name "Sam Harris Ohio University"
```
